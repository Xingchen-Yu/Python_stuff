## Adaptive

```python
import numpy as np
import random

# Simulate the actual lake fish density (ground truth for simulation purposes)
np.random.seed(0)  # For reproducibility
lake_size = (10, 10)
lake = np.random.randint(0, 20, size=lake_size)  # Random fish counts in each cell (hidden to the sampler)

# Initial random samples (10 random samples)
initial_samples = [(random.randint(0, 9), random.randint(0, 9)) for _ in range(10)]
initial_fish_counts = [lake[loc] for loc in initial_samples]

# Set to track visited cells
visited = set(initial_samples)

# Adaptive sampling: focus on areas with high density (threshold)
threshold = 10  # If initial sample fish count is >= 10, we adaptively sample neighbors
adaptive_samples = []
total_sampled_fish = sum(initial_fish_counts)

for loc in initial_samples:
    if lake[loc] >= threshold:  # If fish count exceeds threshold, sample neighbors
        x, y = loc
        neighbors = [(i, j) for i in range(max(0, x-1), min(10, x+2)) 
                     for j in range(max(0, y-1), min(10, y+2)) if (i, j) not in visited]
        adaptive_samples.extend(neighbors)
        visited.update(neighbors)  # Mark neighbors as visited

# Sample fish counts at adaptive sample locations
adaptive_fish_counts = [lake[loc] for loc in adaptive_samples]
total_sampled_fish += sum(adaptive_fish_counts)

# Estimate total fish population in the lake
total_samples = len(initial_samples) + len(adaptive_samples)
lake_area = lake_size[0] * lake_size[1]
estimated_total_fish = (total_sampled_fish / total_samples) * lake_area

# Output the results
print("Initial Samples:", initial_samples)
print("Adaptive Samples:", adaptive_samples)
print("Total Sampled Fish:", total_sampled_fish)
print(f"Estimated Total Fish Population: {estimated_total_fish:.2f}")
print(f"Actual Total Fish Population (for comparison): {lake.sum()}")  # Hidden in real-world cases
```

## Horvitz–Thompson estimator (PPS)

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x = np.random.poisson(lam = 5, size=10000)
```

```python
y = np.random.randint(low=10,high=50, size=500)
```

```python
population = np.concatenate((x,y))
```

```python
df = pd.DataFrame(population,columns=['usage'])
df['usage_group'] = pd.cut(df['usage'],bins=[-1,5, 10, 30, np.inf],labels=[1,2,3,4])
```

```python
df['strata_weights'] = df.groupby('usage_group')['usage_group'].transform('size') / len(df)
```

```python
df['strata_weights'] = df['strata_weights']/df['strata_weights'].sum()
```
