```python
# Convert a column to datetime
df['date_column'] = pd.to_datetime(df['date'])
df
```

```python
df['year_month'] = df['date_column'].dt.strftime('%Y-%m')
df
```

```python
df['year'] = df['date_column'].dt.year
df['month'] = df['date_column'].dt.month
df['day'] = df['date_column'].dt.day
df['weekday'] = df['date_column'].dt.weekday  # Monday = 0, Sunday = 6
df['quarter'] = df['date_column'].dt.quarter
df
```

```python
df = df.sort_values(['date_column','year','weekday']).reset_index(drop=True)
df['diff_days'] = df.groupby(['year','weekday'])['date_column'].diff().dt.days
df
```

```python
df['date_plus_10_days'] = df['date_column'] + pd.Timedelta(days=10)
df
```

```python
df['date_diff'] = df['date_plus_10_days'] - df['date_column']
df
```

```python
df['diff_days'].isna().sum()
```

```python
df['diff_days'] = df['diff_days'].fillna(pd.Timestamp('2020-01-01')) # df['diff_days'].fill_na(pd.Timestamp('2020-01-01'), inplace=True)
df
```

```python
df['previous_day'] = df['value'].shift(1)  # Shift by 1 row (previous day)
df['next_day'] = df['value'].shift(-1)  # Shift forward by 1 row (next day)
df
```

```python
import pandas as pd
from datetime import datetime, timedelta

# Assuming you have a DataFrame 'df' with a date column 'date_column'
# First, ensure the date column is in datetime format
df['date_column'] = pd.to_datetime(df['date_column'])

# Get the current date
current_date = datetime.now()

# Calculate the date one month ago
one_month_ago = current_date - pd.DateOffset(months=1)

# Filter the DataFrame to get rows with dates in the last month
last_month_data = df[df['date_column'] > one_month_ago]

# If you want just the unique dates within the last month
unique_dates_last_month = last_month_data['date_column'].dt.date.unique()

print(unique_dates_last_month)
```

```python
import pandas as pd
from datetime import datetime

# Assuming you have a DataFrame 'df' with a date column 'date_column'
df['date_column'] = pd.to_datetime(df['date_column'])

# Get the current date
current_date = pd.Timestamp.now()

# Calculate the date 30 days ago
thirty_days_ago = current_date - pd.Timedelta(days=30)

# Filter the DataFrame to get rows with dates in the last 30 days
last_month_data = df[df['date_column'] > thirty_days_ago]
```

__important__
`pd.to_timedelta(merged_df['date_order'], unit='D')` allows for series operation comparing the `pd.Timedelta` and `pd.Dateoffset`

```python
def active_users(accounts: pd.DataFrame, logins: pd.DataFrame) -> pd.DataFrame:
    merged_df = logins.drop_duplicates().merge(accounts, on ='id')
    merged_df['login_date'] = pd.to_datetime(merged_df['login_date'])
    merged_df.sort_values(by=['id','login_date'], inplace=True)
    merged_df['date_order'] = merged_df.groupby(['id'])['login_date'].transform('rank',method='dense')
    merged_df['diff_group'] = merged_df['login_date'] - pd.to_timedelta(merged_df['date_order'], unit='D')
    return merged_df.groupby(['id','name','diff_group']).size().to_frame(name='counts').reset_index().query('counts>=5')[['id','name']].drop_duplicates()
```
